### Bài 6: `Bai_6.py`

```
a,b,c=map(float,input().split())
if (a<0) or (b<0) or (c<0):
    print("Không phải là 3 cạnh của tam giác")
elif (a+b>c) and (a+c>b) and (b+c>a):
    p=(a+b+c)/2
    S=(p*(p-a)*(p-b)*(p-c))**0.5
    print(f"{S:.1f}")
else:
    print("Không phải là 3 cạnh của tam giác")
```

### Bài 17: `Bai_khac_9.py`

```
def check_number(x):
    match (x):
        case 0:
            return "Không"
        case 1:
            return "Một"
        case 2:
            return "Hai"
        case 3:
            return "Ba"
        case 4:
            return "Bốn"
        case 5:
            return "Năm"
        case 6:
            return "Sáu"
        case 7:
            return "Bảy"
        case 8:
            return "Tám"
        case 9:
            return "Chín"
        case _:
            return "Không phải số có một chữ số"
n=int(input())
print(check_number(n))

            
```
